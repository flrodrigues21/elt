# 01 - Python para Dados

 conceitos fundamentais de Python usados em pipelines de dados.

## 1. Colecoes

```python
# Listas
cidades = ['Sao Paulo', 'Rio', 'Belo Horizonte']
print(cidades[0])
print(cidades[-1:])
```

```python
# Dicionarios
funcionario = {'nome': 'Ana', 'depto': 'TI', 'salario': 9500}
print(funcionario['nome'])
```

```python
# Comprehensions
quadrados = [x**2 for x in range(10)]
print(quadrados)

salarios = {k: v for k, v in [('Ana', 9500), ('Joao', 7200), ('Maria', 11000)] if v > 8000}
print(salarios)
```

## 2. Funcoes e Lambda

```python
funcionarios = [
    {'nome': 'Ana', 'salario': 9500},
    {'nome': 'Joao', 'salario': 7200},
    {'nome': 'Maria', 'salario': 11000},
]

aumento = lambda f: {**f, 'salario': f['salario'] * 1.10}
novos = list(map(aumento, funcionarios))
print(novos)
```

## 3. Tratamento de Erros

```python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return 'Divisao por zero!'
    except TypeError as e:
        return f'Tipo invalido: {e}'

print(dividir(10, 3))
print(dividir(10, 0))
print(dividir(10, 'x'))
```

## 4. Arquivos

```python
import json, tempfile, os
caminho = os.path.join(tempfile.gettempdir(), 'exemplo.json')
with open(caminho, 'w') as f:
    json.dump({'nome': 'ELT Lab', 'versao': 1}, f)

with open(caminho) as f:
    dados = json.load(f)
print(dados)
os.remove(caminho)
```

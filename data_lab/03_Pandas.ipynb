# 03 - Pandas

Pandas e a ferramenta principal para manipulacao de dados em Python.

## 1. DataFrame

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'nome': ['Ana', 'Joao', 'Maria', 'Pedro', 'Lucia'],
    'depto': ['TI', 'RH', 'TI', 'Financeiro', 'RH'],
    'salario': [9500, 7200, 11000, 8800, 6500],
    'admissao': pd.to_datetime(['2020-01-15', '2019-06-01', '2021-03-10', '2018-11-20', '2022-08-05'])
})
print(df)
print(df.dtypes)
```

## 2. Filtros

```python
ti = df[df['depto'] == 'TI']
print(ti)

alta_rem = df[(df['salario'] > 8000) & (df['depto'] == 'TI')]
print(alta_rem)
```

## 3. GroupBy

```python
resumo = df.groupby('depto').agg(
    qtd=('nome', 'count'),
    salario_medio=('salario', 'mean'),
    salario_max=('salario', 'max'),
).round(2)
print(resumo)
```

## 4. Merge

```python
departamentos = pd.DataFrame({
    'depto': ['TI', 'RH', 'Financeiro'],
    'localizacao': ['Sao Paulo', 'Rio', 'Belo Horizonte']
})

merged = df.merge(departamentos, on='depto', how='left')
print(merged)
```

## 5. Tratamento de NaN

```python
df_nan = pd.DataFrame({'A': [1, np.nan, 3], 'B': [np.nan, 2, 3]})
print(df_nan)
print('---')
print(df_nan.fillna(0))
print('---')
print(df_nan.dropna())
```

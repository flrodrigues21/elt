Pipeline completo de Machine Learning supervisionado.

## 1. Dados Sinteticos

````python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=500, n_features=8, n_informative=5,
                           n_classes=2, random_state=42)

df = pd.DataFrame(X, columns=[f'feat_{i}' for i in range(8)])
df['target'] = y
print(df.head())
print(f'Distribuicao: {dict(pd.Series(y).value_counts())}')
````

## 2. Train/Test Split

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.drop('target', axis=1), df['target'],
    test_size=0.2, random_state=42, stratify=df['target']
)
print(f'Treino: {len(X_train)}, Teste: {len(X_test)}')
````

## 3. Pipeline

````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(random_state=42, max_iter=1000)),
])

pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
print(f'Acuracia: {accuracy_score(y_test, y_pred):.4f}')
print()
print(classification_report(y_test, y_pred))
````

## 4. Prevencao de Data Leakage

**Regra de ouro**: sempre faca `train_test_split` ANTES de qualquer transformacao.
O Pipeline garante que `StandardScaler` so enxerga dados de treino.


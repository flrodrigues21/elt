# 08 - Visualizacao de Dados

Matplotlib, Seaborn e Plotly para analise exploratoria.

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
df = pd.DataFrame({
    'departamento': rng.choice(['TI', 'RH', 'Vendas', 'Financeiro'], 200),
    'salario': rng.integers(4000, 15000, 200),
    'satisfacao': rng.uniform(3, 10, 200).round(1),
    'anos': rng.integers(1, 20, 200),
})
```

## 1. Matplotlib - Subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(df['salario'], bins=20, color='steelblue', edgecolor='white')
axes[0].set_title('Distribuicao de Salarios')
axes[0].set_xlabel('Salario')

depto_mean = df.groupby('departamento')['salario'].mean().sort_values()
depto_mean.plot(kind='barh', ax=axes[1], color='coral')
axes[1].set_title('Salario Medio por Departamento')

plt.tight_layout()
plt.savefig('/tmp/matplotlib_exemplo.png', dpi=100)
plt.show()
print('Salvo em /tmp/matplotlib_exemplo.png')
```

## 2. Seaborn

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=df, x='salario', hue='departamento', ax=axes[0], kde=True)
axes[0].set_title('Salario por Departamento')

sns.heatmap(df[['salario', 'satisfacao', 'anos']].corr(), annot=True, cmap='coolwarm', ax=axes[1])
axes[1].set_title('Correlacao')

plt.tight_layout()
plt.show()
```

## 3. Plotly - Interativo

```python
fig = px.scatter(df, x='salario', y='satisfacao', color='departamento',
                 hover_data=['anos'], title='Salario vs Satisfacao')
fig.write_html('/tmp/plotly_exemplo.html')
fig.show()
print('Salvo em /tmp/plotly_exemplo.html')
```

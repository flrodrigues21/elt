# 02 - NumPy

NumPy e a base para computacao numerica em Python.

## 1. Criando Arrays

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
print(a)
print(f'Tipo: {type(a)}, Shape: {a.shape}, Dtype: {a.dtype}')
```

```python
zeros = np.zeros((3, 4))
print(zeros)

linhas = np.arange(0, 100, 5)
print(linhas)
```

```python
rng = np.random.default_rng(42)
matriz = rng.standard_normal((4, 4))
print(matriz.round(2))
```

## 2. Indexacao e Fatiamento

```python
v = np.array([10, 20, 30, 40, 50, 60])
print(f'v[2]:      {v[2]}')
print(f'v[1:4]:    {v[1:4]}')
print(f'v[::2]:    {v[::2]}')
print(f'v[::-1]:   {v[::-1]}')
```

```python
m = np.arange(12).reshape(3, 4)
print(m)
print(f'm[1, 2]:   {m[1, 2]}')
print(f'm[:, 0]:   {m[:, 0]}')
```

## 3. Broadcasting

```python
a = np.array([[1], [2], [3]])
b = np.array([10, 20, 30, 40])
print(a + b)
```

## 4. Agregacoes

```python
dados = rng.standard_normal(1000)
print(f'Media:        {dados.mean():.4f}')
print(f'Desvio padrao: {dados.std():.4f}')
print(f'Min:          {dados.min():.4f}')
print(f'Max:          {dados.max():.4f}')
print(f'P25:          {np.percentile(dados, 25):.4f}')
print(f'P75:          {np.percentile(dados, 75):.4f}')
```

## 5. Operacoes Vetorizadas

```python
x = np.array([1, 4, 9, 16, 25], dtype=float)
print(f'Raiz:     {np.sqrt(x)}')
print(f'Log:      {np.log(x).round(3)}')
print(f'Soma cum: {np.cumsum(x)}')
```

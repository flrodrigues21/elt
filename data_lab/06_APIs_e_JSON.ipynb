# 06 - APIs e JSON

Como consumir e testar APIs REST com Python.

## 1. Mock Server Local

```python
import threading, json
from http.server import HTTPServer, BaseHTTPRequestHandler

class Handler(BaseHTTPRequestHandler):
    PRODUTOS = [
        {'id': 1, 'nome': 'Notebook', 'preco': 4500},
        {'id': 2, 'nome': 'Mouse', 'preco': 89},
        {'id': 3, 'nome': 'Teclado', 'preco': 199},
        {'id': 4, 'nome': 'Monitor', 'preco': 1200},
        {'id': 5, 'nome': 'Headset', 'preco': 350},
    ]
    def do_GET(self):
        self.send_response(200)
        self.send_header('Content-Type', 'application/json')
        self.end_headers()
        self.wfile.write(json.dumps(self.PRODUTOS).encode())
    def log_message(self, *a): pass

server = HTTPServer(('127.0.0.1', 15555), Handler)
th = threading.Thread(target=server.serve_forever, daemon=True)
th.start()
print('Mock server rodando em http://127.0.0.1:15555')
```

## 2. GET simples

```python
import requests

resp = requests.get('http://127.0.0.1:15555')
print(f'Status: {resp.status_code}')
dados = resp.json()
print(f'Produtos: {len(dados)}')
for p in dados:
    print(f"  {p['nome']}: R$ {p['preco']}")
```

## 3. Tratamento de Erros

```python
try:
    resp = requests.get('http://127.0.0.1:15555/rota_inexistente', timeout=3)
    resp.raise_for_status()
except requests.exceptions.HTTPError as e:
    print(f'HTTP Error: {e}')
except requests.exceptions.ConnectionError as e:
    print(f'Connection Error: {e}')
```

## 4. JSON Normalization

```python
from pandas import json_normalize

df_api = json_normalize(dados)
print(df_api)
```

## 5. Encerramento

```python
server.shutdown()
print('Servidor encerrado')
```
